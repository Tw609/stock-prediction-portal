## SUBPLOTS AND LEGENDS

```python
days = [1,2,3,4,5]
original_price = [10,20,30,40,50]
prediction_price = [10,20,20,100,120]
```

```python
plt.plot(days,original_price,'b-o', label='Original Price')
plt.plot(days,prediction_price,'r-o', label='Predicted Price')
plt.legend()
plt.title('Plot with legend');
plt.xlabel('Weekend');
plt.ylabel('Price');
```

## SUBPLOTS

```python
# FIRST SUBPLOT - ORIGINAL PRICE
plt.subplot(3,1,1)
plt.plot(days,original_price,'b-o', label='Original Price')
plt.title('Original prices over time');
plt.xlabel('Weekend');
plt.ylabel('Price');
plt.legend()

# SECOND SUBPLOT - PREDICTED PRICE
plt.subplot(3,1,2)
plt.plot(days,prediction_price,'r-o', label='Predicted Price')
plt.title('Predicted prices over time');
plt.xlabel('Weekend');
plt.ylabel('Price');
plt.legend()

# DIFFERENCE BETWEEN ORIGINAL & PREDICTED PRICES
price_different = [o - p for o, p in zip(original_price, prediction_price)]
plt.subplot(3,1,3)
plt.bar(days,prediction_price, color='purple', label='Price Different');
plt.title('Different Between Original & Predict Price');
plt.xlabel('Days');
plt.ylabel('Price Different');
plt.legend()

plt.tight_layout()
```

## SAVING THE FIGURE

```python
plt.plot(days,original_price,'b-o', label='Original Price')
plt.plot(days,prediction_price,'r-o', label='Predicted Price')
plt.legend()
plt.title('Plot with legend');
plt.xlabel('Weekend');
plt.ylabel('Price');
plt.savefig('example_prediction.png', dpi=369) #SAVING THE IMAGE
```
